### Files 
In this notebook, we will cover...
- saving progress in a game in a file
- loading game progress

```python
import pygame
import os

SAVEFILE = "gameprogress.txt"

def save_progress(score):
    """Write the score to a file."""
    with open(SAVEFILE, "w") as f:
        f.write(str(score))
    print("Progress saved!")

def load_progress():
    """Read the score from file if it exists."""
    if os.path.exists(SAVEFILE):
        with open(SAVEFILE, "r") as f:
            return int(f.read())
    return 0   # default if no file yet

def display_score(screen, score):
    '''Display the score on the screen'''
    font_small = pygame.font.SysFont(None, 28)
    str_score = str(score)
    s = font_small.render(str_score, True, (230, 230, 230))
    srect = s.get_rect(topleft=(10, 10))
    screen.blit(s, srect)

def main():
    pygame.init()
    screen = pygame.display.set_mode((400, 300))
    pygame.display.set_caption("Save Example")
    clock = pygame.time.Clock()

    score = load_progress()

    

    running = True
    
    # game loop
    while running:
        dt = clock.tick(60)

        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False

            # If the user presses x, the progress is saved
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_x:
                    save_progress(score)

        # here's a test that this is working
        score += 1   # the score goes up each iteration
        if score % 10 == 0:
            display_score(screen, score)
            

        # draw on screen
        screen.fill((30, 30, 30))       # clear frame
        display_score(screen, score)    # draw ON TOP of background
        pygame.display.flip()           # show it


    pygame.quit()


if __name__ == "__main__":
    main()
```
